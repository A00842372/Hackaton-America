## 11. Descargar todos los recursos de las temporadas de Liga MX

La función siguiente recorre cada temporada de Liga MX disponible y guarda un archivo JSON por recurso, partido o temporada en `datos_statsbomb/`. Omite archivos ya descargados, de modo que se puede reanudar. Una descarga completa requiere miles de peticiones y bastante espacio; ejecútala expresamente con `descargar_todo()`. Para comprobarla primero con un partido, usa `descargar_todo(max_partidos=1)`.


```python
def descargar_todo(max_partidos=None, destino="datos_statsbomb"):
    """Guarda todos los recursos accesibles sin acumularlos en memoria."""
    raiz = Path(destino)
    raiz.mkdir(parents=True, exist_ok=True)
    recursos_partido = {
        "alineaciones": sb.lineups,
        "eventos": sb.events,
        "frames_360": sb.frames,
        "jugadores_partido": sb.player_match_stats,
        "equipos_partido": sb.team_match_stats,
    }
    recursos_temporada = {
        "jugadores_temporada": sb.player_season_stats,
        "equipos_temporada": sb.team_season_stats,
    }
    errores = []
    descargados = 0
    partidos_procesados = 0

    def guardar_json(ruta, obtener):
        nonlocal descargados
        if ruta.exists():
            return
        ruta.parent.mkdir(parents=True, exist_ok=True)
        try:
            datos = obtener()
            temporal = ruta.with_suffix(".json.tmp")
            with temporal.open("w", encoding="utf-8") as archivo:
                json.dump(datos, archivo, ensure_ascii=False)
            temporal.replace(ruta)
            descargados += 1
        except Exception as exc:
            errores.append((str(ruta), f"{type(exc).__name__}: {exc}"))

    # Solo Liga MX varonil (competition_id 73).
    competiciones_liga_mx = competiciones.loc[competiciones["competition_id"].eq(73)]
    guardar_json(
        raiz / "competiciones.json",
        lambda: competiciones_liga_mx.to_dict("records"),
    )
    for fila in competiciones_liga_mx.itertuples(index=False):
        if max_partidos is not None and partidos_procesados >= max_partidos:
            break
        cid, sid = int(fila.competition_id), int(fila.season_id)
        carpeta = raiz / f"competicion_{cid}" / f"temporada_{sid}"
        try:
            lista_partidos = sb.matches(competition_id=cid, season_id=sid, creds=credenciales)
        except Exception as exc:
            errores.append((str(carpeta / "partidos.json"), f"{type(exc).__name__}: {exc}"))
            continue
        guardar_json(carpeta / "partidos.json", lambda p=lista_partidos: p.to_dict("records"))
        for nombre, funcion in recursos_temporada.items():
            guardar_json(
                carpeta / f"{nombre}.json",
                lambda f=funcion, c=cid, s=sid: f(competition_id=c, season_id=s, fmt="json", creds=credenciales),
            )
        for mid in lista_partidos["match_id"].astype(int):
            if max_partidos is not None and partidos_procesados >= max_partidos:
                print(f"Archivos nuevos: {descargados}; partidos visitados: {partidos_procesados}; errores: {len(errores)}")
                return errores
            for nombre, funcion in recursos_partido.items():
                guardar_json(
                    carpeta / nombre / f"{mid}.json",
                    lambda f=funcion, m=mid: f(match_id=m, fmt="json", creds=credenciales),
                )
            partidos_procesados += 1
    print(f"Archivos nuevos: {descargados}; partidos visitados: {partidos_procesados}; errores: {len(errores)}")
    return errores

# Ejecuta una de estas líneas cuando quieras guardar los datos:
# errores = descargar_todo(max_partidos=1)  # prueba pequeña
# errores = descargar_todo()                # descarga completa
```

```python
# 13. Descarga paralela de partidos, eventos, estadísticas de equipo, alineaciones y 360 de Liga MX.
# Ajusta max_trabajadores (predeterminado: 4). Los archivos completos se reutilizan al reanudar.

from concurrent.futures import ThreadPoolExecutor, as_completed
import time


def descargar_bases_todas(destino="datos_statsbomb", max_trabajadores=4):
    if max_trabajadores < 1:
        raise ValueError("max_trabajadores debe ser al menos 1")
    raiz = Path(destino)
    recursos = {
        "eventos": sb.events,
        "equipos_partido": sb.team_match_stats,
        "alineaciones": sb.lineups,
        "frames_360": sb.frames,
    }
    catalogo = sb.competitions(creds=credenciales)
    # Solo Liga MX varonil (competition_id 73).
    catalogo = catalogo.loc[catalogo["competition_id"].eq(73)]
    if catalogo.empty:
        raise RuntimeError("La API no devolvió temporadas de Liga MX; no se inició la descarga.")

    raiz.mkdir(parents=True, exist_ok=True)
    df_partidos_temporadas = []
    inventario = []
    errores = []

    def guardar(ruta, datos):
        ruta.parent.mkdir(parents=True, exist_ok=True)
        temporal = ruta.with_suffix(".json.tmp")
        try:
            with temporal.open("w", encoding="utf-8") as archivo:
                json.dump(datos, archivo, ensure_ascii=False)
            temporal.replace(ruta)
        finally:
            temporal.unlink(missing_ok=True)

    def descargar_uno(nombre, funcion, mid, ruta):
        for intento in range(3):
            try:
                datos = funcion(match_id=mid, fmt="json", creds=credenciales)
                if not datos:
                    return "respuesta_vacia", None
                guardar(ruta, datos)
                return "descargado", None
            except Exception as exc:
                if intento == 2:
                    return "error", f"{type(exc).__name__}: {exc}"
                time.sleep(2 ** intento)

    for comp in catalogo.drop_duplicates(["competition_id", "season_id"]).itertuples(index=False):
        cid, sid = int(comp.competition_id), int(comp.season_id)
        carpeta = raiz / f"competicion_{cid}" / f"temporada_{sid}"
        print(f"Consultando {comp.competition_name} {comp.season_name}...")
        try:
            partidos_temporada = sb.matches(
                competition_id=cid, season_id=sid, creds=credenciales)
            if partidos_temporada.empty:
                raise ValueError("La API devolvió una lista de partidos vacía")
            guardar(carpeta / "partidos.json", partidos_temporada.to_dict("records"))
        except Exception as exc:
            errores.append({"competition_id": cid, "season_id": sid,
                            "recurso": "partidos", "error": str(exc)})
            print(f"  No se pudo consultar la temporada: {type(exc).__name__}")
            continue

        df_partidos_temporadas.append(partidos_temporada)
        tareas = []
        for partido in partidos_temporada.itertuples(index=False):
            mid = int(partido.match_id)
            for nombre, funcion in recursos.items():
                disponible = (partido.match_status_360 == "available"
                              if nombre == "frames_360"
                              else partido.match_status == "available")
                ruta = carpeta / nombre / f"{mid}.json"
                info = {"competition_id": cid, "season_id": sid,
                        "temporada": comp.season_name, "match_id": mid,
                        "local": partido.home_team, "visitante": partido.away_team,
                        "recurso": nombre, "archivo": str(ruta)}
                if not disponible:
                    inventario.append({**info, "estado": "no_disponible_en_catalogo"})
                elif ruta.is_file() and ruta.stat().st_size > 2:
                    inventario.append({**info, "estado": "ya_descargado"})
                else:
                    tareas.append((info, nombre, funcion, mid, ruta))

        ejecutor = ThreadPoolExecutor(max_workers=max_trabajadores)
        try:
            futuros = {
                ejecutor.submit(descargar_uno, nombre, funcion, mid, ruta): info
                for info, nombre, funcion, mid, ruta in tareas
            }
            for indice, futuro in enumerate(as_completed(futuros), start=1):
                info = futuros[futuro]
                estado, error = futuro.result()
                inventario.append({**info, "estado": estado})
                if error:
                    errores.append({**info, "error": error})
                if indice % 100 == 0 or indice == len(tareas):
                    print(f"  {indice}/{len(tareas)} recursos consultados")
        finally:
            ejecutor.shutdown(wait=True, cancel_futures=True)
        print(f"  {len(partidos_temporada)} partidos revisados")

    df_partidos_todos = (pd.concat(df_partidos_temporadas, ignore_index=True)
                         .drop_duplicates("match_id").reset_index(drop=True)
                         if df_partidos_temporadas else pd.DataFrame())
    return df_partidos_todos, pd.DataFrame(inventario), pd.DataFrame(errores)


df_partidos, df_inventario_descarga, df_errores_descarga = descargar_bases_todas()
print(f"Partidos en df_partidos: {len(df_partidos)}; errores: {len(df_errores_descarga)}")
if not df_inventario_descarga.empty:
    display(df_inventario_descarga.groupby(["temporada", "recurso", "estado"])
            .size().rename("partidos").reset_index())
if not df_errores_descarga.empty:
    display(df_errores_descarga)
```

```python
# 12. Prueba de velocidad: un partido nuevo por temporada de Liga MX.
# Guarda la muestra en datos_statsbomb para que la descarga completa pueda reutilizarla.

from concurrent.futures import ThreadPoolExecutor, as_completed
from time import perf_counter


def probar_descarga_muestra(destino="datos_statsbomb", max_trabajadores=4):
    """Descarga un partido nuevo por temporada de Liga MX y mide los cuatro recursos."""
    raiz = Path(destino)
    recursos = {
        "eventos": sb.events,
        "equipos_partido": sb.team_match_stats,
        "alineaciones": sb.lineups,
        "frames_360": sb.frames,
    }
    catalogo = sb.competitions(creds=credenciales)
    catalogo = catalogo.loc[catalogo["competition_id"].eq(73)]
    if catalogo.empty:
        raise RuntimeError("No se encontraron temporadas de Liga MX")

    def guardar(ruta, datos):
        ruta.parent.mkdir(parents=True, exist_ok=True)
        temporal = ruta.with_suffix(".json.tmp")
        try:
            with temporal.open("w", encoding="utf-8") as archivo:
                json.dump(datos, archivo, ensure_ascii=False)
            temporal.replace(ruta)
        finally:
            temporal.unlink(missing_ok=True)

    inicio_total = perf_counter()
    catalogos = []
    tareas = []
    for comp in catalogo.drop_duplicates(["competition_id", "season_id"]).itertuples(index=False):
        cid, sid = int(comp.competition_id), int(comp.season_id)
        partidos_temporada = sb.matches(
            competition_id=cid, season_id=sid, creds=credenciales)
        carpeta = raiz / f"competicion_{cid}" / f"temporada_{sid}"
        guardar(carpeta / "partidos.json", partidos_temporada.to_dict("records"))
        candidatos = partidos_temporada.loc[
            partidos_temporada["match_status"].eq("available")
            & partidos_temporada["match_status_360"].eq("available")
        ]
        # Una ejecución posterior elige otro partido si este ya se descargó.
        elegido = None
        for partido in candidatos.itertuples(index=False):
            mid = int(partido.match_id)
            if all(not (carpeta / nombre / f"{mid}.json").exists()
                   for nombre in recursos):
                elegido = mid
                break
        catalogos.append({
            "temporada": comp.season_name,
            "partidos": len(partidos_temporada),
            "eventos_disponibles": int(partidos_temporada["match_status"].eq("available").sum()),
            "360_disponibles": int(partidos_temporada["match_status_360"].eq("available").sum()),
            "match_id_muestra": elegido,
        })
        if elegido is not None:
            for nombre, funcion in recursos.items():
                tareas.append((comp.season_name, elegido, nombre, funcion,
                               carpeta / nombre / f"{elegido}.json"))

    def descargar(tarea):
        temporada, mid, nombre, funcion, ruta = tarea
        inicio = perf_counter()
        try:
            datos = funcion(match_id=mid, fmt="json", creds=credenciales)
            segundos_api = perf_counter() - inicio
            if not datos:
                return {"temporada": temporada, "match_id": mid, "recurso": nombre,
                        "estado": "respuesta_vacia", "bytes": 0,
                        "segundos_api": segundos_api}
            guardar(ruta, datos)
            return {"temporada": temporada, "match_id": mid, "recurso": nombre,
                    "estado": "descargado", "bytes": ruta.stat().st_size,
                    "segundos_api": segundos_api}
        except Exception as exc:
            return {"temporada": temporada, "match_id": mid, "recurso": nombre,
                    "estado": f"error: {type(exc).__name__}", "bytes": 0,
                    "segundos_api": perf_counter() - inicio}

    inicio_archivos = perf_counter()
    with ThreadPoolExecutor(max_workers=max_trabajadores) as ejecutor:
        futuros = [ejecutor.submit(descargar, tarea) for tarea in tareas]
        archivos = [futuro.result() for futuro in as_completed(futuros)]
    segundos_archivos = perf_counter() - inicio_archivos
    segundos_total = perf_counter() - inicio_total
    return (pd.DataFrame(catalogos), pd.DataFrame(archivos),
            segundos_archivos, segundos_total)


df_catalogos_prueba, df_archivos_prueba, segundos_archivos, segundos_total = probar_descarga_muestra()
display(df_catalogos_prueba)
if df_archivos_prueba.empty:
    print("No quedan partidos sin descargar para esta prueba.")
else:
    display(df_archivos_prueba.groupby(["recurso", "estado"]).agg(
        archivos=("match_id", "size"),
        megabytes=("bytes", lambda x: x.sum() / 1_000_000),
        segundos_api_promedio=("segundos_api", "mean"),
    ).reset_index())
    exitos = df_archivos_prueba.loc[df_archivos_prueba["estado"].eq("descargado")]
    promedios = exitos.groupby("recurso")["bytes"].mean()
    n_eventos = int(df_catalogos_prueba["eventos_disponibles"].sum())
    n_360 = int(df_catalogos_prueba["360_disponibles"].sum())
    conteos = {"eventos": n_eventos, "equipos_partido": n_eventos,
               "alineaciones": n_eventos, "frames_360": n_360}
    if all(recurso in promedios for recurso in conteos):
        bytes_estimados = sum(promedios[recurso] * cantidad
                              for recurso, cantidad in conteos.items())
        horas_transferencia_20_mbps = bytes_estimados / 2_500_000 / 3600
        print(f"Tamaño estimado: {bytes_estimados / 1e9:.1f} GB; "
              f"transferencia mínima a 20 Mbps: {horas_transferencia_20_mbps:.1f} h")
    print(f"Tiempo de los archivos: {segundos_archivos:.1f} s; "
          f"total con catálogos: {segundos_total:.1f} s")
```

## 14. DataFrames para análisis

Este bloque reúne las tablas para análisis. `df_partidos` carga todos los archivos `partidos.json` guardados en `datos_statsbomb`, de cualquier competición y temporada. Las demás tablas corresponden a las consultas anteriores: las de temporada usan `competition_id` y `season_id`, y las de partido usan `match_id`. `df_alineaciones` une ambos equipos en una sola tabla. `bases_df` permite acceder a todas por nombre.


```python
df_competiciones = competiciones
archivos_partidos = sorted(Path("datos_statsbomb").glob("competicion_*/temporada_*/partidos.json"))
df_partidos = pd.DataFrame(
    partido
    for archivo in archivos_partidos
    for partido in json.loads(archivo.read_text(encoding="utf-8"))
)
if not df_partidos.empty:
    df_partidos = df_partidos.drop_duplicates(subset="match_id").reset_index(drop=True)
df_alineaciones = (
    pd.concat(
        [tabla.assign(equipo=equipo) for equipo, tabla in alineaciones.items()],
        ignore_index=True,
    )
    if alineaciones else pd.DataFrame()
)
df_eventos = eventos
df_frames_360 = frames_360
df_jugadores_partido = estadisticas_jugadores_partido
df_equipos_partido = estadisticas_equipos_partido
df_jugadores_temporada = estadisticas_jugadores_temporada
df_equipos_temporada = estadisticas_equipos_temporada

bases_df = {
    "competiciones": df_competiciones,
    "partidos": df_partidos,
    "alineaciones": df_alineaciones,
    "eventos": df_eventos,
    "frames_360": df_frames_360,
    "jugadores_partido": df_jugadores_partido,
    "equipos_partido": df_equipos_partido,
    "jugadores_temporada": df_jugadores_temporada,
    "equipos_temporada": df_equipos_temporada,
}

resumen_bases = pd.DataFrame(
    [{"base": nombre, "filas": len(tabla), "columnas": len(tabla.columns)}
     for nombre, tabla in bases_df.items()]
)
display(resumen_bases)
```
