# Datos de StatsBomb

Ejecuta las celdas en orden. La cuenta tiene acceso a varias temporadas de Liga MX y Liga MX Femenil. Cada bloque importa un recurso de la API como DataFrame o diccionario de DataFrames.


## 0. Librerías y credenciales

Las credenciales están en `.env`, excluido de Git. Usa el kernel **Python (Hackaton-America)**. [Referencia del cliente oficial](https://github.com/hudl/statsbombpy/blob/master/README.md).


```python
import json
import os
from pathlib import Path

import numpy as np
import pandas as pd
import requests
import sklearn
from dotenv import load_dotenv

load_dotenv(".env")
usuario = os.getenv("API_USERNAME")
contrasena = os.getenv("API_PASSWORD")
if not usuario or not contrasena:
    raise ValueError("Completa API_USERNAME y API_PASSWORD en .env")
credenciales = {"user": usuario, "passwd": contrasena}

from statsbombpy import sb

print(f"pandas {pd.__version__} | numpy {np.__version__} | scikit-learn {sklearn.__version__}")
```

## 1. Competiciones y temporadas

`competiciones` contiene todas las combinaciones disponibles para esta cuenta. Los identificadores `competition_id` y `season_id` sirven para las siguientes consultas.


```python
competiciones = sb.competitions(creds=credenciales)
display(competiciones[["competition_id", "season_id", "competition_name", "season_name"]])
```

## 2. Partidos de una temporada

El ejemplo usa Liga MX 2025/2026. Cambia los identificadores según la tabla anterior.


```python
competition_id = 73
season_id = 318
partidos = sb.matches(competition_id=competition_id, season_id=season_id, creds=credenciales)
display(partidos[["match_id", "match_date", "home_team", "away_team"]].head())
```

## 3. Elegir un partido

Los siguientes bloques consultan el mismo `match_id`. Puedes cambiarlo por cualquier identificador de `partidos`.


```python
match_id = int(partidos.iloc[0]["match_id"])
print("Partido seleccionado:", match_id)
```

## 4. Alineaciones

El resultado es un diccionario: cada equipo tiene su propio DataFrame.


```python
alineaciones = sb.lineups(match_id=match_id, creds=credenciales)
for equipo, tabla in alineaciones.items():
    print(equipo, len(tabla), "jugadores")
    display(tabla.head())
```

## 5. Eventos

Pases, tiros y el resto de acciones del partido en un DataFrame.


```python
eventos = sb.events(match_id=match_id, creds=credenciales)
print(len(eventos), "eventos")
display(eventos.head())
```

## 6. Datos 360

Posiciones de jugadores visibles y áreas visibles asociadas a eventos, si están disponibles. Esta tabla puede ser grande.


```python
frames_360 = sb.frames(match_id=match_id, creds=credenciales)
print(len(frames_360), "filas 360")
display(frames_360.head())
```

## 7. Estadísticas de jugadores por partido


```python
estadisticas_jugadores_partido = sb.player_match_stats(match_id=match_id, creds=credenciales)
print(len(estadisticas_jugadores_partido), "jugadores")
display(estadisticas_jugadores_partido.head())
```

## 8. Estadísticas de equipos por partido


```python
estadisticas_equipos_partido = sb.team_match_stats(match_id=match_id, creds=credenciales)
print(len(estadisticas_equipos_partido), "equipos")
display(estadisticas_equipos_partido.head())
```

## 9. Estadísticas de jugadores por temporada


```python
estadisticas_jugadores_temporada = sb.player_season_stats(
    competition_id=competition_id, season_id=season_id, creds=credenciales
)
print(len(estadisticas_jugadores_temporada), "jugadores")
display(estadisticas_jugadores_temporada.head())
```

## 10. Estadísticas de equipos por temporada


```python
estadisticas_equipos_temporada = sb.team_season_stats(
    competition_id=competition_id, season_id=season_id, creds=credenciales
)
print(len(estadisticas_equipos_temporada), "equipos")
display(estadisticas_equipos_temporada.head())
```

## 14. DataFrames para análisis

Este bloque reúne las tablas para análisis. `df_partidos` carga todos los archivos `partidos.json` guardados en `datos_statsbomb`, de cualquier competición y temporada. Las demás tablas corresponden a las consultas anteriores: las de temporada usan `competition_id` y `season_id`, y las de partido usan `match_id`. `df_alineaciones` une ambos equipos en una sola tabla. `bases_df` permite acceder a todas por nombre.


```python
df_competiciones = competiciones
archivos_partidos = sorted(Path("datos_statsbomb").glob("competicion_*/temporada_*/partidos.json"))
df_partidos = pd.DataFrame(
    partido
    for archivo in archivos_partidos
    for partido in json.loads(archivo.read_text(encoding="utf-8"))
)
if not df_partidos.empty:
    df_partidos = df_partidos.drop_duplicates(subset="match_id").reset_index(drop=True)
df_alineaciones = (
    pd.concat(
        [tabla.assign(equipo=equipo) for equipo, tabla in alineaciones.items()],
        ignore_index=True,
    )
    if alineaciones else pd.DataFrame()
)
df_eventos = eventos
df_frames_360 = frames_360
df_jugadores_partido = estadisticas_jugadores_partido
df_equipos_partido = estadisticas_equipos_partido
df_jugadores_temporada = estadisticas_jugadores_temporada
df_equipos_temporada = estadisticas_equipos_temporada

bases_df = {
    "competiciones": df_competiciones,
    "partidos": df_partidos,
    "alineaciones": df_alineaciones,
    "eventos": df_eventos,
    "frames_360": df_frames_360,
    "jugadores_partido": df_jugadores_partido,
    "equipos_partido": df_equipos_partido,
    "jugadores_temporada": df_jugadores_temporada,
    "equipos_temporada": df_equipos_temporada,
}

resumen_bases = pd.DataFrame(
    [{"base": nombre, "filas": len(tabla), "columnas": len(tabla.columns)}
     for nombre, tabla in bases_df.items()]
)
display(resumen_bases)
```
